Date Standardization:

Business Metric Creation:
Derived financial metrics are created from the original sales and cost variables. Revenue, cost, gross profit, and gross margin are calculated to support profitability analysis.

```python
df["revenue"] = df["units_sold"] * df["unit_price"]

df["cost"] = df["units_sold"] * df["unit_cost"]

df["gross_profit"] = df["revenue"] - df["cost"]
```

Forecast Performance Metrics:
Forecast error and absolute forecast error are calculated to measure the difference between actual sales and demand forecasts. Forecast accuracy is calculated as a percentage, with undefined values retained as missing when the demand forecast is zero.

```python
df["forecast_error"] = df["units_sold"] - df["demand_forecast"]

df["absolute_forecast_error"] = df["forecast_error"].abs()

# Gross Margin %
df["gross_margin_pct"] = np.where(
    df["revenue"] > 0,
    (df["gross_profit"] / df["revenue"]) * 100,
    np.nan
)

# Forecast Accuracy %
df["forecast_accuracy_pct"] = np.where(
    df["demand_forecast"] > 0,
    np.maximum(0,
    (1 - df["absolute_forecast_error"] / df["demand_forecast"]) * 100
    ),
    np.nan
)
```

```python
print("Zero demand forecast:",
      (df["demand_forecast"] == 0).sum())

print("Missing forecast accuracy:",
      df["forecast_accuracy_pct"].isna().sum())

print("Zero demand forecast + non-missing accuracy:",
      ((df["demand_forecast"] == 0) &
       (df["forecast_accuracy_pct"].notna())).sum())
```

Inventory Performance Metrics:
Inventory gap measures the difference between current inventory and the reorder point. A replenishment risk flag is created when inventory falls at or below the reorder point. Inventory coverage estimates the number of days that available inventory can support sales based on the current sales rate.

```python
df["inventory_gap"] = df["inventory_level"] - df["reorder_point"]

df["replenishment_risk"] = np.where(
    df["inventory_level"] <= df["reorder_point"],
    1,
    0
)
```

```python
df["inventory_coverage_days"] = np.where(
    df["units_sold"] > 0,
    df["inventory_level"] / df["units_sold"],
    np.nan
)
```

```python
df["year"] = df["date"].dt.year

df["month"] = df["date"].dt.month

df["month_name"] = df["date"].dt.month_name()

df["day_of_week"] = df["date"].dt.day_name()

df["day_of_week_num"] = df["date"].dt.dayofweek
```
